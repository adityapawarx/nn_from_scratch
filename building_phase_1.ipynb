```python
#Lets code Sigmoid, ReLU (Rectified Linear Unit), GELU and their derivatives in Python

import numpy as np
def sigmoid(x):
    """Compute the sigmoid activation function."""
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    """Compute the derivative of the sigmoid activation function."""
    s = sigmoid(x)
    return s * (1 - s)

def relu(x):
    """Compute the ReLU activation function."""
    return np.maximum(0, x)

def relu_derivative(x):
    """Compute the derivative of the ReLU activation function."""
    return (x > 0).astype(float)

def gelu(x):
    """Compute the GELU activation function."""
    return 0.5 * x * (1 + np.tanh(np.sqrt(2 / np.pi) * (x + 0.044715 * x**3)))

def gelu_derivative(x):
    """Compute the derivative of the GELU activation function."""
    c = np.sqrt(2 / np.pi)
    x_cubed = x**3
    tanh_arg = c * (x + 0.044715 * x_cubed)
    tanh_val = np.tanh(tanh_arg)
    sech_squared = 1 - tanh_val**2
    return 0.5 * (1 + tanh_val + x * c * 0.044715 * 3 * x**2 * sech_squared)
```
